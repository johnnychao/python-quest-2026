```python
#@title 🧰 魔法工具箱：先在右邊填「暱稱」，再按左邊的 ▶ 執行這一格 { display-mode: "form" }
暱稱 = "" #@param {type:"string"}
# ======================================================================
#  Python 冒險之旅 2026 · 關卡檢查工具（看不懂沒關係，這一格不是今天的功課 😉）
# ======================================================================
import hashlib, unicodedata, io, sys, re, contextlib, traceback, builtins

_LEVEL = "L02"
_SALT = "python-quest-2026-datama"
_TASKS = ["2-1", "2-2", "2-3", "2-4", "2-5"]
_XP_EACH = 20
_CHECKS = {}
_PASSED = builtins.__dict__.setdefault("_pyquest_" + _LEVEL, {})
_HINTS = {}

def _norm_name(s):
    return re.sub(r"\s+", "", unicodedata.normalize("NFKC", str(s))).lower()

def _squash(s):
    return re.sub(r"\s+", "", str(s))

def 出現(out, *subs):
    """輸出中是否（忽略空白）包含所有片段"""
    o = _squash(out)
    return all(_squash(x) in o for x in subs)

def 數字們(out):
    """抓出輸出裡所有的數字（float）"""
    return [float(x) for x in re.findall(r"-?\d+(?:\.\d+)?", str(out))]

def 行列表(out):
    return [ln.rstrip() for ln in str(out).splitlines() if ln.strip()]

class _NeedMoreInput(Exception):
    pass

_HIST = builtins.__dict__.setdefault("_pyquest_hist", [])
def _on_pre_run(*args):
    try:
        info = args[0]
        src = getattr(info, "raw_cell", None)
        if isinstance(src, str):
            _HIST.append(src)
    except Exception:
        pass
try:
    _ip = get_ipython()
    if not builtins.__dict__.get("_pyquest_hooked"):
        _ip.events.register("pre_run_cell", _on_pre_run)
        builtins.__dict__["_pyquest_hooked"] = True
except Exception:
    pass

def _history():
    try:
        ip = get_ipython()
        h = list(ip.user_ns.get("In") or ip.user_ns.get("_ih") or [])
    except Exception:
        h = list(globals().get("In") or [])
    return [c for c in (h + list(_HIST)) if isinstance(c, str)]

def _find_cell(tid):
    marker = "# 🎯 任務 " + tid
    for cell in reversed(_history()):
        if marker in cell:
            lines = [ln for ln in cell.splitlines()
                     if not re.match(r"\s*(檢查|通關密語)\s*\(", ln)]
            return "\n".join(lines)
    return None

def _make_runner(src):
    def run(*inputs):
        feed = iter([str(x) for x in inputs])
        buf = io.StringIO()
        try:
            ns = dict(get_ipython().user_ns)
        except Exception:
            ns = dict(globals())
        def _fake_input(prompt=""):
            try:
                return next(feed)
            except StopIteration:
                raise _NeedMoreInput()
        ns["input"] = _fake_input
        ns["__name__"] = "__main__"
        try:
            import matplotlib.pyplot as _plt
            _plt.close("all"); _orig_show = _plt.show; _plt.show = lambda *a, **k: None
        except Exception:
            _plt = None
        try:
            with contextlib.redirect_stdout(buf):
                exec(compile(src, "<任務 " + _LEVEL + ">", "exec"), ns)
        finally:
            if _plt is not None:
                _plt.show = _orig_show
        return buf.getvalue(), ns
    run.src = src
    return run

def 任務定義(tid, fn, 提示=""):
    _CHECKS[tid] = fn
    _HINTS[tid] = 提示

def _progress():
    done = sum(1 for t in _TASKS if _PASSED.get(t))
    bar = "■" * done + "□" * (len(_TASKS) - done)
    return f"[{bar}] {done}/{len(_TASKS)}"

def 檢查(tid):
    tid = str(tid)
    if tid not in _CHECKS:
        print(f"⚠️ 找不到任務 {tid} 的檢查設定。"); return
    src = _find_cell(tid)
    if src is None:
        print(f"❌ 找不到「# 🎯 任務 {tid}」的程式格。請先執行那一格（並保留第一行的標記），再執行這裡。")
        return
    run = _make_runner(src)
    try:
        result = _CHECKS[tid](run)
    except _NeedMoreInput:
        result = (False, "你的程式呼叫 input() 的次數比題目預期的多，請檢查輸入的次數。")
    except Exception as e:
        tb = traceback.format_exc().strip().splitlines()[-1]
        result = (False, f"程式執行時發生錯誤 → {tb}")
    ok, extra = (result, "") if isinstance(result, bool) else result
    if ok:
        first = not _PASSED.get(tid)
        _PASSED[tid] = True
        print(f"✅ 任務 {tid} 通過！{'+' + str(_XP_EACH) + ' XP ' if first else ''}{_progress()}")
        if all(_PASSED.get(t) for t in _TASKS):
            print("🏆 本關所有任務都完成了！請執行最下面的「通關密語」那一格。")
    else:
        print(f"❌ 任務 {tid} 還沒通過。{_progress()}")
        if extra: print("   💬 " + str(extra))
        if _HINTS.get(tid): print("   💡 提示：" + _HINTS[tid])
        print("   👉 修改程式後，先重新執行任務那一格，再執行這一格。")

def 通關密語():
    missing = [t for t in _TASKS if not _PASSED.get(t)]
    if missing:
        print("🔒 還有任務未通過：" + "、".join(missing) + "　完成後再來拿密語吧！")
        return
    name = 暱稱.strip() if isinstance(暱稱, str) else ""
    if not name:
        name = input("請輸入你在入口網頁登錄的暱稱：").strip()
    if not name:
        print("⚠️ 暱稱不能是空白。"); return
    code = hashlib.sha256(f"{_SALT}|{_LEVEL}|{_norm_name(name)}".encode("utf-8")).hexdigest()[:6].upper()
    print("=" * 46)
    print(f"🎉 恭喜 {name}！{_LEVEL} 通關！")
    print(f"🔑 通關密語：PYQ-{_LEVEL}-{code}")
    print("👉 回到入口網頁，把密語貼到這一關的「輸入通關密語」欄位。")
    print("=" * 46)

print(f"🧰 魔法工具箱已準備好！本關有 {len(_TASKS)} 個任務：{'、'.join(_TASKS)}")
print("   做完每個任務後，執行它下方的「檢查」格；全部通過後執行最下方的「通關密語」。")

# ---------------- 各任務的檢查規則 ----------------
def _check_2_1(run):
    out, ns = run()
    if not isinstance(ns.get("name"), str): return (False, "name 應該是字串（要加引號）。")
    if not isinstance(ns.get("level"), int) or isinstance(ns.get("level"), bool): return (False, "level 應該是整數 1。")
    if not isinstance(ns.get("hp"), float): return (False, "hp 應該是浮點數 100.0（要有小數點）。")
    return (ns["level"] == 1 and ns["hp"] == 100.0, "level 要是 1，hp 要是 100.0。")
任務定義("2-1", _check_2_1, 提示="字串要加引號；整數不加小數點；浮點數要有小數點。")

def _check_2_2(run):
    out, ns = run()
    o = _squash(out)
    need = ["<class'int'>", "<class'float'>", "<class'str'>", "<class'bool'>"]
    if not all(n in o for n in need): return (False, "應該要出現 int、float、str、bool 四種型別。")
    return (o.count("<class'bool'>") >= 2, "True 和 99 > 1 都是 bool，所以 bool 要出現兩次。")
任務定義("2-2", _check_2_2, 提示="type() 的括號裡放值，再用 print() 包起來印出。")

def _check_2_3(run):
    out, ns = run()
    src = run.src
    if 'a = "奶茶"' in src or "a = '奶茶'" in src or 'b = "咖啡"' in src or "b = '咖啡'" in src:
        return (False, "不可以直接重新指定字串，要用交換的寫法。")
    return (ns.get("a") == "奶茶" and ns.get("b") == "咖啡", "交換後 a 應該是奶茶、b 是咖啡。")
任務定義("2-3", _check_2_3, 提示="課本 2.3：a, b = b, a 可以同時交換兩個變數。")

def _check_2_4(run):
    out, ns = run()
    want = {"n": 124, "f": 25.0, "i": 3, "r": 4, "t": "100元"}
    for k, v in want.items():
        if ns.get(k) != v or type(ns.get(k)) != type(v):
            return (False, f"變數 {k} 應該是 {v!r}（型別 {type(v).__name__}），現在是 {ns.get(k)!r}。")
    return True
任務定義("2-4", _check_2_4, 提示="int('123') → 123；float('12.5') → 12.5；int(3.99) → 3；round(3.99) → 4；str(100) → '100'。")

def _check_2_5(run):
    out, ns = run()
    nums = 數字們(out)
    return (90 in nums and 80 in nums and "A" in out, "應該印出 90 80 A。")
任務定義("2-5", _check_2_5, 提示="不能以數字開頭、不能有減號、不能用保留字 class；可以用底線，例如 first_score。")
```

## 📦 2-1　變數：幫資料貼標籤的盒子
- `=` 是**指定運算子**：把右邊的值放進左邊的變數。`x = 5` 念作「把 5 指定給 x」。
- 變數名稱（識別字）規則：只能用**英文字母、數字、底線**；不能以數字開頭；不能用 Python 保留字（如 `if`、`for`、`class`）；大小寫不同（`Score` 和 `score` 是兩個變數）。
- 好習慣：用有意義的英文名稱，例如 `total_price`、`user_name`。

```python
yesNo = 'y'                                # 字串
passScore, maxScore, minScore = 60, 100, 0 # 一次指定三個變數（課本 ex02/variable.py）
power1 = power2 = 100                      # 兩個變數同一個值
total = 1.23456E+6                         # 科學記號 → 1234560.0
print(yesNo)
print(passScore, maxScore, minScore)
print(power1, power2)
print(total)
```

## 2-2　基本資料型別：Python 怎麼看待資料
| 型別 | 例子 | 說明 |
|---|---|---|
| `int` 整數 | `12`, `-3`, `0b1100`(二進位) | 沒有小數點 |
| `float` 浮點數 | `3.14`, `1.2e3` | 有小數點或科學記號 |
| `str` 字串 | `'Python'`, `"3.8"` | 引號包起來的文字，`"3.8"` 是字串不是數字！ |
| `bool` 布林 | `True`, `False` | 只有真／假兩種值，注意大寫開頭 |

用 `type()` 可以問 Python「這是什麼型別？」

```python
print(type(12), type(3.14), type("Python"), type(True))
print(type("False"))    # 有引號 → 字串
print(type(2.0))        # 有小數點 → 浮點數
print(type(+2E12))      # 科學記號 → 浮點數（課本 test02_01.py）
print(bool(0), bool(1), bool(""), bool("Python"))   # 什麼會被當成 False？
```

## 2-3　型別轉換：`int()`、`float()`、`str()`、`round()`
從鍵盤或檔案讀進來的資料通常是**字串**，要計算前必須先轉成數字。

```python
i1 = 10
f1 = float(i1)          # 整數 → 浮點數 10.0
f2 = 1234.5678
i2 = int(f2)            # 浮點數 → 整數（直接捨去小數）1234
i3 = round(f2)          # 四捨五入 1235
s = str(i2)             # 整數 → 字串 '1234'
print(f1, i2, i3, s, type(s))
print(round(f2, 2))     # 四捨五入到小數第 2 位
print(int("123") + 1)   # 字串 '123' → 整數 123，再加 1
print("123" + "1")      # 字串 + 字串 → 接在一起！
```

## 🔍 小知識：`id()` 與物件
Python 的資料都是**物件**，`id()` 會顯示物件在記憶體的位置（課本 2.1.2）。

```python
x = 5
y = x
print(id(x), id(y))   # 兩個名字指向同一個物件
x = 3 + y
print(id(x), id(y))   # x 換了新物件，y 不變
```
