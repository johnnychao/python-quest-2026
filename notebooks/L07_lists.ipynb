## 7-2　串列的方法與內建函式（課本 6.4）
| 方法／函式 | 功能 |
|---|---|
| `lst.append(x)` | 在最後加入 x |
| `lst.insert(i, x)` | 在索引 i 插入 x |
| `lst.remove(x)` | 移除第一個 x |
| `lst.pop()` / `lst.pop(i)` | 取出並移除最後一個／索引 i |
| `lst.index(x)`、`lst.count(x)` | 找索引、算次數 |
| `lst.sort()`、`lst.reverse()` | 就地排序／反轉（**會改變串列**） |
| `sorted(lst)` | 傳回新的排序串列，原串列不變 |
| `len()`、`sum()`、`max()`、`min()` | 長度、總和、最大、最小 |
| `x in lst` | 是否包含 |

```python
score = [72, 98, 86, 76, 63]
score.append(90); print(score)
score.insert(0, 100); print(score)
score.remove(63); print(score)
last = score.pop(); print(last, score)
score.sort(); print(score)                  # 由小到大
score.sort(reverse=True); print(score)      # 由大到小
print(len(score), sum(score), max(score), min(score))
animal = ['dog', 'cat', 'monkey']
print(sorted(animal), animal)               # sorted 不改原串列
print('cat' in animal, animal.index('cat'))
```

## 7-3　走訪串列：`for…in` 與 `for…range`

```python
season = ['spring', 'summer', 'autumn', 'winter']
for item in season:                         # 直接拿元素
    print(item, end=' ')
print()
for i in range(len(season)):                # 用索引（需要位置時）
    print(i, season[i])
for i, item in enumerate(season, start=1):  # enumerate：同時拿到編號與元素
    print(f'第{i}季：{item}')
```
